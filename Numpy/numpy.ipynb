### Vector, Matrix and Tensor

```python
vector=np.array([1,2,3])
print("vector : ", vector)

matrix= np.array([[1,2,3],
                [4,5,6]])
print("Matrix : ", matrix)

tensor=np.array([[[1,2], [3,4]],
               [[5,6], [7,8]]])
print("tensor : ", tensor)
```

### Array Properties

```python
arr =np.array([[1,2,3],
              [4,5,6]])

print("Shape", arr.shape)
print("Dimension", arr.ndim)
print("Size", arr.size)
print("DType", arr.dtype)
```

### Array Reshaping

```python
arr=np.arange(12)
print("original array :", arr)

reshape=arr.reshape((3,4))
print("reshape array :", reshape)

flattend=reshape.flatten()
print("fllatten array :", flattend)  #it returns the cooy of the array

raveled=reshape.ravel()
print("reveled array :", raveled)  #it returns the view instead of copy

#Transpose
transpose=reshape.T
print("Transpose array :", transpose)
```
