### Array Reshaping

```python
arr=np.arange(12)
print("original array :", arr)

reshape=arr.reshape((3,4))
print("reshape array :", reshape)

flattend=reshape.flatten()
print("fllatten array :", flattend)  #it returns the cooy of the array

raveled=reshape.ravel()
print("reveled array :", raveled)  #it returns the view instead of copy

#Transpose
transpose=reshape.T
print("Transpose array :", transpose)
```

## Phase - 3, Advance operation with Business examples

```python
#Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data=np.array([
    [1,150000, 180000, 220000, 250000],       #Paradise Biriyani
    [2, 120000, 140000, 160000, 190000],      #Beijing Bites
    [3, 200000, 230000, 260000, 300000],      #Pizza Hub
    [4, 160000, 185000, 205000, 230000],      #Chai Point
    [5, 183000, 120000, 170000, 100000]       #Burger King
])
```

```python
import matplotlib.pyplot as plt
```

```python
print("=== zomato sales analysis === ")
print("\n Sales data shape", sales_data.shape)
print("\n Sample data for 1st 3 renstau: ", sales_data[:, 1:])
```

```python
#total sales per year

print(np.sum(sales_data, axis=0))
yearly_total=np.sum(sales_data[:, 1:], axis=0)
print(yearly_total)
```

```python
#Minimun Sales per restaurant

min_sales=np.min(sales_data[:, 1:], axis=1)   #0 for updown and 1 of sidewise
print(min_sales)
```

```python
#Maximum sales per year

max_sales=np.max(sales_data[:, 1:], axis= 0)
print(max_sales)
```

```python
#Avergage sales for restaurant

avg_sales=np.average(sales_data[:, 1:], axis=1)
print(avg_sales)
```

```python
#Cummulative  Sales

cumsum=np.cumsum(sales_data[:, 1:], axis=1)
print(cumsum)
```

```python
#showing in graph
plt.figure(figsize=(10,6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average cummalative sales across all restaurant")
plt.xlabel('Years')
plt.ylabel("sales")
plt.grid(True)
plt.show()
```
