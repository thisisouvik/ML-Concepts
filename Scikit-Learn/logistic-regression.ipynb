## Data Engineering

```python
df['family']= df['sibsp'] + df['parch']
df.loc[df['family'] > 0, 'travelled_alone'] = 0
df.loc[df['family']== 0, 'travelled_alone'] = 1
```

```python
df['family'].head()
```

```python
df.drop(['parch', 'sibsp'], axis= 1, inplace= True)
```

```python
sns.countplot(x='travelled_alone', data= df)
plt.title("No. of apssebngers travelling alone")
plt.show()
```

```python
df.head()
```

```python
df.drop(['name', 'ticket', 'home.dest'], axis= 1, inplace= True)
```

```python
df.head()
```

```python
df.drop(['body',], axis= 1, inplace= True)
df.head()
```

```python
sex = pd.get_dummies(df['sex'], drop_first= True)
print(sex)
```

```python
df.loc(df['sex']== 'True', 'gender') = 0
df.loc(df['sex']== 'False', 'gender') = 1
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.drop(['cabin', 'boat'], axis= 1, inplace= True)
```

```python
df.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
imp_mean = SimpleImputer(strategy= 'mean')
```

```python
df['age'] = imp_mean.fit_transform(df[['age']])
df['fare'] = imp_mean.fit_transform(df[['fare']])
```

```python
df.isnull().sum()
```

```python
imp_freq = SimpleImputer(strategy= 'most_frequent')
```

```python
df['embarked'] = imp_freq.fit_transform(df[['embarked']])
```

```python
df.head()
```

```python
embark =pd.get_dummies(df['embarked'], drop_first= True)
print(embark)
```

```python
df.drop(['embarked'], axis= 1, inplace= True)
df = pd.concat([df,embark], axis= 1)
```

```python
df.head()
```

```python
X=df.drop(['survived'], axis=1)
X.head()
```

```python
y= df['survived']
y.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.3, random_state= 1)
```

```python
print(X_train.shape, y_train.shape)
```

```python
print(X_test.shape, y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
mod= LogisticRegression()
mod.fit(X_train, y_train)
```

```python
pred = mod.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(pred, y_test)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test, pred)          #Truepositive  Falsepositive
                                        #False Negetive  True negetive
```
