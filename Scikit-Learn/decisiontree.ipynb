```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder()

df['BP'] = oe.fit_transform(df['BP'])
df['Sex'] = oe.fit_transform(df['Sex'])
df['Cholestrol'] = oe.fit_transform(df['Cholestrol'])
df['Drug'] = oe.fit_transform(df['Drug'])
```

```python
X=df.iloc[:, 0:-1]
y=df.iloc[:, -1]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state= 42, test_size= 0.25)
```

```python
from sklearn.tree import DecisionTreeClassifier

ds=DecisionTreeClassifier(criterion='gini', random_state=0)
ds.fit(X_train, y_train)
predict=ds.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(predict, y_test)
```

```python
from sklearn import tree
plt.figure(figsize=(10,10))
tree.plot_tree(ds.fit(X_train,y_train))
plt.show()
```

```python
clf_entrophy= DecisionTreeClassifier(criterion='entropy', max_depth=3, random_state=0)
clf_entrophy.fit(X_train,y_train)
predict2=clf_entrophy.predict(X_test)
```

```python
accuracy_score(y_test, predict2)
plt.figure(figsize=(10,10))
tree.plot_tree(ds.fit(X_train,y_train))
plt.show()
```
