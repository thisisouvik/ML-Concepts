## Exploring Pandas Library in this notebook

```python
#Before using it Pandas Module should be installed

import pandas as pd
```

## Importing Datase using Pandas in my Text Editor

```python
data=pd.read_csv("dataset.csv")
```

```python
#printing head of the data

print(data.head(5))
```

## Reading Data in Pandas

```python
#Reading Header

data.columns
```

```python
#Reading Particular index by name

print(data[['title', 'genre', 'popularity']])  #slicing 0:5

#or you can just do data.title but above is more accurate
```

```python
#printing one row 

data.iloc[1:3]
```

```python
#give me details of paticular index like here the "title"

data.iloc[2,1]
```

```python
#going to specific row or property

data.loc[data['genre']== "Drama"] #restriction only one property at a time
```

# SUMMARY

for acessing dataset Pandas and just -> pd.csv_read("<name.type>")

for printing column -> data.column

for acessing rows -> data.iloc[slicing]

for locating specific object of dataset for -> data.loc

## Sorting Accending/Decending

```python
data.describe()
```

```python
#sorting

data.sort_values(['release_date', 'title'], ascending=[1,0]) #firstly ascending and then decending
```

# Making Changes in dataset

```python
data['Total'] = data['id'] + data['popularity']   #making another cloumn  making a total

data=data.drop['Total'] #deleting another column completeply
```
