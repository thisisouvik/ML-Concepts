## Sorting Accending/Decending

```python
data.describe()
```

```python
#sorting

data.sort_values(['id', 'title'], ascending=[1,0]) #firstly ascending and then decending
```

# Making Changes in dataset

```python
data['Total'] = data['id'] + data['popularity']   #making another cloumn  making a total

data=data.drop(columns=['Total']) #deleting another column completeply
```

# Filtering our Dataset

```python
new_data=data.loc[(data['genre']== 'Crime') & (data['original_language']== 'en')]   #and = & or= | we can do conditions like & data['vote' > 6.9] 
```

```python
new_data.reset_index(drop=True, inplace=True)

new_data
```

```python
#selecting particular string inside a object in column

data.loc[~data['title'].str.contains("Good")]  #delete directly by selecting "~"
```

```python
data.loc[data['title'].str.contains("Th[a-z]*")]   #filtering based on the first 2 letter
```
