```python
x=[1,2,3]
y=[4,5,6]
plt.plot(x,y)
plt.grid(True)
plt.show()
```

```python
data={
    "Salary": [25000 , 30000, 28000, 48000, 52000, 27000 ]
}
```

```python
df=pd.DataFrame(data)
```

```python
df.head()
df.shape
```

```python
plt.plot(df["Salary"], color = 'red', marker= "o", linestyle = "--")
plt.grid(True)
plt.show()
```

### Histogram

```python

plt.hist(df['Salary'])
plt.show()
```

```python


plt.hist(df['Salary'])
plt.show()
```

### Boxplot

```python

plt.boxplot(df["Salary"])
plt.show()
```

```python
df["dept"] = ['HR', 'IT', 'Finance', 'HR', 'Security','HR']
df.head()
```

```python
#Pie Chart

count=df["dept"].value_counts()
plt.pie(count, labels= count.index, autopct="%1.2f", explode= [0, 0.1, 0.2, 0.1])
plt.axis('equal')
plt.title("Dept and employee")
plt.show()
```

### Count Plot

```python
plt.bar(count.index, count, color= ['green', 'red', 'black'])
plt.show()
        
```

```python
#Count.plot

plt.bar(count.index, count, color= ['green', 'red', 'black'])
plt.show()
        
```

### Bivarite - numerical - numerical analysis


```python
df["Age"]= [22,21,45,65,33,29]
df.head()
```

### Scatter Plot

```python
plt.scatter(df["Age"], df["Salary"], color = "black")
plt.show()
```

```python
#Line plot

sort_age= df.sort_values("Age")

plt.plot(sort_age["Age"], df["Salary"], color = "red", marker = ".", linewidth = "2")
plt.grid(True)
plt.show()
```

```python
# Bar chart: 
plt.bar(sort_age["Age"], df["Salary"], color= "Green", align="center")
```

```python
#Bivarite analysis: Numerical - categorical
hr_sal= df[df["dept"] == "HR"]["Salary"]
it_sal= df[df["dept"] == "IT"]["Salary"]
```

```python
plt.boxplot([hr_sal, it_sal], tick_labels = ["HR", "IT"])
plt.show()
```

```python
#Piechart

salary_by_dept = df.groupby("dept")["Salary"].sum()
salary_by_dept
```

```python
#pie_chart
plt.pie(salary_by_dept, labels= salary_by_dept.index, autopct= "%1.2f")
plt.axis('Equal')
plt.show()
```

```python
#Bar plot

hr_mean=sum(hr_sal)/len(hr_sal)
it_mean=sum(it_sal)/len(it_sal)
plt.bar(["HR", "IT"], [hr_mean, it_mean], color=["red", "yellow"])
plt.show()
```
