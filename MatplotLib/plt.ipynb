# Matplot Library in Python

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
x=[1,2,3]
y=[4,5,6]
plt.plot(x,y)
plt.grid(True)
plt.show()
```

```python
data={
    "Salary": [25000 , 30000, 28000, 48000, 52000, 27000 ]
}
```

```python
df=pd.DataFrame(data)
```

```python
df.head()
df.shape
```

```python
plt.plot(df["Salary"], color = 'red', marker= "o", linestyle = "--")
plt.grid(True)
plt.show()
```

```python
#Histogram

plt.hist(df['Salary'])
plt.show()
```

```python
#Boxplot

plt.boxplot(df["Salary"])
plt.show()
```

```python
df["dept"] = ['HR', 'IT', 'Finance', 'HR', 'Security','HR']
df.head()
```

```python
#Pie Chart

count=df["dept"].value_counts()
plt.pie(count, labels= count.index, autopct="%1.2f", explode= [0, 0.1, 0.2, 0.1])
plt.axis('equal')
plt.title("Dept and employee")
plt.show()
```

```python
#Count.plot

plt.bar(count.index, count, color= ['green', 'red', 'black'])
plt.show()
        
```
