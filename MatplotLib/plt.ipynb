```python
data={
    "Salary": [25000 , 30000, 35000, 48000, 52000, 60000 ]
}
```

```python
df["dept"] = ['HR', 'IT', 'Finance', 'HR', 'Security','HR']
df.head()
```

```python
#Pie Chart

count=df["dept"].value_counts()
plt.pie(count, labels= count.index, autopct="%1.2f", explode= [0, 0.1, 0.2, 0.1])
plt.axis('equal')
plt.title("Dept and employee")
plt.show()
```

### Count Plot

```python
plt.bar(count.index, count, color= ['green', 'red', 'black'])
plt.show()
        
```

```python
#Count.plot

plt.bar(count.index, count, color= ['green', 'red', 'black'])
plt.show()
        
```

### Bivarite - numerical - numerical analysis


```python
df["Age"]= [22,21,45,65,33,29]
df.head()
```

### Scatter Plot

```python
plt.scatter(df["Age"], df["Salary"], color = "black")
plt.show()
```

```python
#Line plot

sort_age= df.sort_values("Age")

plt.plot(sort_age["Age"], df["Salary"], color = "red", marker = ".", linewidth = "2")
plt.grid(True)
plt.show()
```

```python
# Bar chart: 
plt.bar(sort_age["Age"], df["Salary"], color= "Green", align="center")
```

```python
#Bivarite analysis: Numerical - categorical
hr_sal= df[df["dept"] == "HR"]["Salary"]
it_sal= df[df["dept"] == "IT"]["Salary"]
```

```python
plt.boxplot([hr_sal, it_sal], tick_labels = ["HR", "IT"])
plt.show()
```

```python
#Piechart

salary_by_dept = df.groupby("dept")["Salary"].sum()
salary_by_dept
```

```python
#pie_chart
plt.pie(salary_by_dept, labels= salary_by_dept.index, autopct= "%1.2f", shadow="True", explode=[0.1,0.1,0.1,0.1])
plt.axis('Equal')
plt.show()
```

```python
#Bar plot

hr_mean=sum(hr_sal)/len(hr_sal)
it_mean=sum(it_sal)/len(it_sal)
plt.bar(["HR", "IT"], [hr_mean, it_mean], color=["red", "yellow"])
plt.show()
```

```python
df['Experince']=[1,2,3,4,5,6]
df.head()
```

```python
#Bubble Plot

plt.scatter(df['Age'], df['Salary'], s=df['Experince']*20, color= "pink", edgecolors='red')
plt.show()
```

```python
# 2 numerical adn 1 catagorical column
plt.scatter(df["Age"], df["Salary"], c=df["dept"].map({'HR': 'yellow', 'IT': 'blue', 'Finance': 'red', 'Security' : 'pink'}))
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary vs Dept")
plt.show()
```

```python
color = {'HR': 'yellow', 'IT': 'blue', 'Finance': 'red', 'Security' : 'pink'}

for dept, color in color.items():
    df_dept = df[df["dept"] == dept]
    plt.scatter(df_dept["Age"], df_dept["Salary"], c= color, label = dept)

plt.legend()
```

### Object Oriented API

```python
fig, axs= plt.subplots(1,3, figsize =(10,5)) # adjust dimension for fitting

#Line Plot
axs[0].plot(sort_age["Age"], df["Salary"], marker='*', linewidth= "2", markersize= 2, color= "red" )
axs[0].grid()
axs[0].set_title("Line plot")
axs[0].set_xlabel("Age")
axs[0].set_ylabel("Salary")

#Histogram

axs[1].hist(df["Salary"], bins=5, color= "skyblue")
axs[1].set_title("Histogram")
axs[1].set_xlabel("Salary")
axs[1].set_ylabel("Frequency")

#Boxplot

axs[2].boxplot(df["Salary"])
axs[2].set_title("Boxplot")
axs[2].set_xlabel("Salary")

plt.savefig("mutipleplots.png") # keeping path in fornt of name will also store in desired location
plt.show()
```

```python
data2= {
    'Year' : [2020, 2021, 2022, 2023],
    'Sales' : [100, 150, 200, 250],
    'Profit' : [20, 30 ,40, 50],
    'Expenses' : [80, 120, 160, 200]
}
```

```python
df2=pd.DataFrame(data2)
```

```python
plt.plot(df2["Year"], df2["Sales"], label= "Sales")
plt.plot(df2["Year"], df2["Profit"], label= "Profit")
plt.plot(df2["Year"], df2["Expenses"], label= "Expenses")

plt.title("Finacial Ananlysis")
plt.xlabel("Year")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python
# 3rd plotting

ax= plt.axes(projection = "3d")
ax.scatter(df["Age"], df["Salary"], df['Experince'])
ax.set_xlabel("Age")
ax.set_ylabel("Salary")
ax.set_zlabel("Experience")
ax.set_title("3D Figure")
plt.show() 
```
