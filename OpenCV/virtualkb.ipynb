```python
import cv2
```

```python
import mediapipe as mp
import cvzone
from cvzone.HandTrackingModule import HandDitector
import time
import keyboard
```

```python
cap= cv2.VideoCapture(0)
cap.set(3,2120)
cap.set(4,1080)

ditector = HandDitector(ditectonCon = 0.8)

def DrawAll(img, buttonList):
    for button in buttonList:
        x, y = button.pos
        w, h = button.size
        cvzone.cornerRect(img, (button.pos[0], button.pos[1], button.size[0], button.size[1]), 20, rt=0)
        cv2.rectangle(img, button.pos, (x+h, y+w), (0,255,255), cv2.FILLED)
        cv2.putText(img, button.text, (x+20, y+65), cv2.FONT_HERSHEY_PLAIN, 4, (255,255,255), 4)
        return img

class Button():
    def __init__(self, pos, text, size= [85,85]):
        self.pos = pos
        self.text = text
        self.size = size

keys = [["Q", "W", "E", "R","..."],["A", "S", "D", "..."],["Z","X","..." "\t"]]

buttonList = []
for i in range(len(keys)):
    for j, key in enumerate(keys[i]):
        buttonList.append(Button([100*j*50, 100*i*50], key))


while True:
    res,img = cap.read()
    img = ditector.findHands(img)
    lmLst , bboxInfo = ditector.findPosition(img)

    img = drawAll(img, buttonList)

    if lmLst:
        for button in buttonList:
            x, y = button.pos
            w, h = button.size

            if x < lmLst[0][0] < x + w and y < lmLst[0][1] < y +h:
                cv2.rectangle(img, (x-5, y-5), (x + w + 5, y + h + 5), (175, 0 , 275), cv2.FILLED)
                cv2.putText(img, button.text, (x+20, y+65), cv2.FONT_HERSHEY_PLAIN, 4, (255,255,255), 4)

                l, _ , _ = ditector.findDistance(0 , 12, img, draw= False)
                print(l)

                if l< 45:
                    keyboard.press(button.text)
                    cv2.rectangle(img, button.pos, (x+w, y+h), (0,0,255), cv2.FILLED)
                    cv2.putText(img, button.test,(x+20, y+20), cv2.FONT_HERSHEY_PLAIN, 4, (255,255,255), 4)

    cv2.imshow(img) 
    if cv2.waitKey(1) & 0xff== ('q'):
        break
```
