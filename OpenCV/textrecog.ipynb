```python
import matplotlib.pyplot as plt 
import numpy as np
```

```python
img_path= "pfp.png"
reader =EasyOCR.Reader(['en'])
result = reader.readtext(img_path)
result
```

```python
top_left= tuple(result[0][0][0])
bottom_right = tuple(result[0][0][1])
text = result[0][1]
```

```python
for ditect in result:
    top_left= tuple(result[0][0][0])
    bottom_right = tuple(result[0][0][1])
    text = ditect[1]
    img = cv2.imread(img_path)
    img = cv2.rectangle(img, top_left, bottom_right, (0,0,255), 2)
    img=cv2.putText(img, top_left, bottom_right, cv2.FONT_HERSHEY_PLAIN, 1, (255,0,0),2)
plt.imshow(img)
plt.show()
```
