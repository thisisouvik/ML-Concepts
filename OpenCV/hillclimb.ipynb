```python
import mediapipe as mp
import pydirectinput
```

```python
import cv2
```

```python
mp_drawing = mp.solutions.drawing_utils
mp_holistic = mp.solutions.holistic
mp_pose= mp.solutions.pose
```

```python
cap = cv2.VideoCapture(0)
cap.set(3,560)
cap.set(4,480)
with mp_holistic.holistic(min_detection_confidence = 0.5, min_tracking_confidence= 0.5) as holistic:

    while cap.isOpened():
        img, sucess = cap.read()
        img = cv2.flip(img, 1)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        result = holistic(img)
        img = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)
        height, width, _= img.shape
        try:
            right_hand = (result.pose_landmark[mp_pose.PoseLandmark.LEFT_WRIST].x * width,
                          result.pose_landmark[mp_pose.PoseLandmark.LEFT_WRIST].y * height)
            left_hand = (result.pose_landmark[mp_pose.PoseLandmark.RIGHT_WRIST].x * width,
                         result.pose_landmark[mp_pose.PoseLandmark.RIGHT_WRIST].y * height)
            y_midline = height//2

            pose = "move"

            if (right_hand[1] < y_midline):
                pose = "acc"
                pydirectinput.keyDown('right')
                pydirectinput.keyUp('left')
            elif(right_hand[1] > y_midline):
                pose = "Brake"
                pydirectinput.keyDown('left')
                pydirectinput.keyUp('right')

        except:
            pass
        cv2.putText(img, pose,(20,0), cv2.FONT_HERSHEY_PLAIN, 2, (255,255,0),2)
        cv2.line(img, (0,y_midline), (width, y_midline), (255,0,0), 2)
        cv2.imshow("Cargame", img)
        if cv2.waitKey(1) & 0xff == ('q'):
            break
cap.release()
cv2.destroyAllWindows()
```
