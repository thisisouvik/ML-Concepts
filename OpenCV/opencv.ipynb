***Colour Detection***

````python
vid3=cv2.VideoCapture(0)
````

````python
#Detecting Red Colour


while True:
    _, img= vid3.read()
    hsv=cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

    redl= np.array([135,85,150], dtype= "uint8")
    redu=np.array([125,250,250], dtype='uint8')
    redM= cv2.inRange(hsv, redl, redu)

    kernel = np.ones([5,5], 'uint8')

    redM = cv2.dilate(redM,kernel)
    red=cv2.bitwise_and(img, img, mask= redM)

    cont, hier = cv2.findContours(redM, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)    #U

    for pic, cont in enumerate (cont):
        area = cv2.contourArea(cont)

        if (area>300):
            x , y, w, h = cv2.boundingRect(cont)
            img = cv2.rectangle(img, (x,y), (x+w, y+h), (0,0,255), 2)
            cv2.putText(img, "red colour", (x,y), cv2.FONT_HERSHEY_PLAIN, 1.0,(0,0,255))
    cv2.imshow("RedColor", img)

    if cv2. waitKey(0) & 0xff == ord('q') :
        cv2.release()
        cv2.destroyAllWindows()
        break
````

***Face Detection***

````python
face_cascade=cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
````

````python
vid4=cv2.VideoCapture(0)
````

````python
while True:
    _ , img= vid4.read()
    gray = cv2.cvtColor(img, cv2.COLOR_BAYER_BG2GRAY)
    faces=face_cascade.detectMultiScale(gray, 1.1, 4) #scale and nearest neighbour
    for x,y,w,h in faces:
        cv2.rectangle(img, (x,y), (255, 0, 0), 4)
    cv2.imshow('Image', img)

    k= cv2.waitKey(1) & 0xff
    if k == ('q'):
        break

vid4.release()
cv2.destroyAllWindows()
````
